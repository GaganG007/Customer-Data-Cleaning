```python
import pandas as pd
```

```python
df = pd.read_excel(r"C:\Users\USER\Desktop\Pandas Tutorial\Customer Call List.xlsx")
df
```

```python
df.drop_duplicates()
```

```python
df["Last_Name"] = df["Last_Name"].str.strip("/_-123..")
```

```python
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace('[^a-zA-Z0-9]','', regex = True)
```

```python
df
```

```python
df["Phone_Number"].apply(lambda x : str(x))
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace('nan','')
df["Phone_Number"] = df["Phone_Number"].str.replace('Na','')
df["Phone_Number"] = df["Phone_Number"].str.replace('NaN','')
df
```

```python
df["Phone_Number"] = df["Phone_Number"].apply(lambda x : str(x))
df["Phone_Number"] = df["Phone_Number"].apply(lambda x : x[0:3] + '-' + x[3:6] + '-' + x[6:10])
df
```

```python
df[["Street Address","State","Zip Code"]] = df["Address"].str.split(',',n = 2, expand = True)
df
```

```python
df = df.drop(columns = [ 'Address', 'Not_Useful_Column'], errors = 'ignore')
```

```python
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace('Yes','Y')
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace('No','N')
df
```

```python
df.replace(['--','N/a','NaN','nan--'],'')
```

```python
df = df[df["Do_Not_Contact"] != 'Y']
df
```

```python
df = df.replace(['--','N/a','NaN','nan--'],'')
df
```

```python
df["Paying Customer"] = df["Paying Customer"].str.replace('Yes','Y')
df["Paying Customer"] = df["Paying Customer"].str.replace('No','N')
df
```

```python
df = df.fillna('')
```

```python
for x in df.index:
    if df.loc[ x, "Phone_Number"] == '':
        df.drop(x , inplace = True)

df
```

```python
df
```

```python
df = df.drop(columns ='Do_Not_Contact')
```
